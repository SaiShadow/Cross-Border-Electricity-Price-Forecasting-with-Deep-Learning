```python
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error

# Load the data and parse the 'Date' column as datetime
data = pd.read_csv('load 2024.csv', parse_dates=['Date'])
data = data.set_index('Date').sort_index()

# Create naive forecast columns for the 'Load' column
data['load_yesterday']  = data['Load'].shift(24)
data['load_last_week']  = data['Load'].shift(24 * 7)
data['load_last_month'] = data['Load'].shift(24 * 30)  # Approximate 30 days

# Drop rows where any of the required values are missing
data.dropna(subset=['Load', 'load_yesterday', 'load_last_week', 'load_last_month'], inplace=True)

# Extract actual and forecasted values
y_actual    = data['Load']
y_yesterday = data['load_yesterday']
y_last_week = data['load_last_week']
y_last_month= data['load_last_month']

# Compute error metrics for the 'yesterday' baseline
mae_yesterday  = mean_absolute_error(y_actual, y_yesterday)
mape_yesterday = mean_absolute_percentage_error(y_actual, y_yesterday)
print(f"Yesterday baseline:  MAE = {mae_yesterday:.2f},  MAPE = {mape_yesterday:.2%}")

# Compute error metrics for the 'last week' baseline
mae_last_week  = mean_absolute_error(y_actual, y_last_week)
mape_last_week = mean_absolute_percentage_error(y_actual, y_last_week)
print(f"Last-week baseline:  MAE = {mae_last_week:.2f},  MAPE = {mape_last_week:.2%}")

# Compute error metrics for the 'last month' baseline
mae_last_month  = mean_absolute_error(y_actual, y_last_month)
mape_last_month = mean_absolute_percentage_error(y_actual, y_last_month)
print(f"Last-month baseline: MAE = {mae_last_month:.2f},  MAPE = {mape_last_month:.2%}")
```
