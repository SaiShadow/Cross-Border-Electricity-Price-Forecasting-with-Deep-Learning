```python
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error

# Load data and set the Date column as index
data = pd.read_csv('renewables.csv', parse_dates=['Date'])
data = data.set_index('Date').sort_index()


# List the target columns you want to forecast
target_columns = ['renewable', 'non_renewable', 'cross_border_trading']

# Create naive forecast columns for each target column
for col in target_columns:
    data[f'{col}_yesterday']   = data[col].shift(24)
    data[f'{col}_last_week']   = data[col].shift(24 * 7)
    data[f'{col}_last_month']  = data[col].shift(24 * 30)  # approximate 30 days

# Evaluate each target's forecast performance
for col in target_columns:
    # For each target, drop rows where the actual or any forecast value is missing
    df = data.dropna(subset=[col, f'{col}_yesterday', f'{col}_last_week', f'{col}_last_month'])
    
    # Extract actual values and forecasts
    y_actual = df[col]
    y_yesterday = df[f'{col}_yesterday']
    y_last_week = df[f'{col}_last_week']
    y_last_month = df[f'{col}_last_month']
    
    # Compute errors for the yesterday baseline
    mae_yesterday = mean_absolute_error(y_actual, y_yesterday)
    mape_yesterday = mean_absolute_percentage_error(y_actual, y_yesterday)
    
    # Compute errors for the last-week baseline
    mae_last_week = mean_absolute_error(y_actual, y_last_week)
    mape_last_week = mean_absolute_percentage_error(y_actual, y_last_week)
    
    # Compute errors for the last-month baseline
    mae_last_month = mean_absolute_error(y_actual, y_last_month)
    mape_last_month = mean_absolute_percentage_error(y_actual, y_last_month)
    
    # Print the results for the current target
    print(f"Forecasting results for '{col}':")
    print(f"  Yesterday baseline:  MAE = {mae_yesterday:.2f},  MAPE = {mape_yesterday:.2%}")
    print(f"  Last-week baseline:  MAE = {mae_last_week:.2f},  MAPE = {mape_last_week:.2%}")
    print(f"  Last-month baseline: MAE = {mae_last_month:.2f},  MAPE = {mape_last_month:.2%}\n")
```
