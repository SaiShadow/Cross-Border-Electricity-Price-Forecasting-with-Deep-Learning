```python
import pandas as pd

# Load the processed electricity price datasets
train_df = pd.read_csv('train_with_monthly_gas.csv')

# Load the gas price dataset (monthly data)
gas_monthly_df = pd.read_csv('gas_2019.csv')  # Replace with the actual file path
gas_monthly_df = gas_monthly_df.rename(columns={"Month": "month", "Gas (NCG, THE)": "gas_price"})

# Add a year column to the gas data
gas_monthly_df['year'] = 2019  # Replace with the corresponding year for each file

# Create a 'date' column representing the first day of each month
gas_monthly_df['date'] = pd.to_datetime(gas_monthly_df[['year', 'month']].assign(day=1))

# Create a temporary 'date' column in the electricity dataset
train_df['date'] = pd.to_datetime(train_df['ds']).dt.to_period('M').dt.to_timestamp()

```

```python
# Merge gas prices into the electricity dataset
train_df = pd.merge(train_df, gas_monthly_df[['date', 'gas_price']], how='left', on='date')

# Drop the temporary 'date' column (it was only used for merging)
train_df = train_df.drop(columns=['date'])

# Save the updated training dataset
train_df.to_csv('train_with_monthly_gas_till_2019.csv', index=False)

print("Monthly gas prices successfully added to the training dataset.")
```
