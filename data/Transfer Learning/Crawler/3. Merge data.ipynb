```python
import pandas as pd
import glob

# List of bidding zones
bidding_zones = [
    "AT", "BE", "CH", "CZ", "DK1", "DK2", "FR", "HU", "IT-North", 
    "NL", "NO2", "PL", "SE4", "SI"
]

# Initialize an empty list to store DataFrames
dfs = []

# Load each CSV file and append to the list
for bidding_zone in bidding_zones:
    file_path = f"{bidding_zone}_training_data_with_calendar.csv"
    try:
        df = pd.read_csv(file_path)
        # df["bidding_zone"] = bidding_zone  # Add a column for bidding zone
        dfs.append(df)
    except FileNotFoundError:
        print(f"File not found for {bidding_zone}, skipping...")

# Concatenate all DataFrames into a single long-format DataFrame
if dfs:
    long_format_df = pd.concat(dfs, ignore_index=True)
    
    # Save the long format DataFrame to a new CSV file
    long_format_df.to_csv("cross_border_electricity_prices.csv", index=False)
    print("Cross-border electricity price dataset saved successfully.")
else:
    print("No data available to merge.")
```
