```python
import pandas as pd

# Load the full dataset for DE-LU
file_path = "DE-LU_training_data_with_calendar.csv"
df_de_lu = pd.read_csv(file_path)

# Convert timestamp column to datetime
df_de_lu["ds"] = pd.to_datetime(df_de_lu["ds"])

# Sort by date to get the most recent data
df_de_lu = df_de_lu.sort_values(by="ds")

# Get the most recent full week (168 rows) + 24 rows of the next day that the model will try to forecast
df_one_shot = df_de_lu.iloc[-192:]

# Get the most recent full month (30 * 24)
df_few_shot = df_de_lu.iloc[-720:]

# Load the cross-border dataset without DE-LU (Zero-Shot)
cross_border_file = "cross_border_electricity_prices.csv"
df_cross_border = pd.read_csv(cross_border_file)

# Remove any existing DE-LU data (if present) for zero-shot learning
df_zero_shot = df_cross_border[df_cross_border["unique_id"] != "DE-LU"]

# Create one-shot dataset by adding the recent DE-LU week
df_one_shot_combined = pd.concat([df_zero_shot, df_one_shot], ignore_index=True)

# Create few-shot dataset by adding the recent DE-LU month
df_few_shot_combined = pd.concat([df_zero_shot, df_few_shot], ignore_index=True)

# Save the datasets
df_zero_shot.to_csv("cross_border_electricity_prices_zero_shot.csv", index=False)
df_one_shot_combined.to_csv("cross_border_electricity_prices_one_shot.csv", index=False)
df_few_shot_combined.to_csv("cross_border_electricity_prices_few_shot.csv", index=False)

print("Datasets for zero-shot, one-shot, and few-shot learning created successfully.")
```
